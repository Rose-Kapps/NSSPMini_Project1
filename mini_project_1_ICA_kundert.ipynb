````python
class DownloadProgressBar(tqdm):
    def update_to(self, b=1, bsize=1, tsize=None):
        if tsize is not None:
            self.total = tsize
        self.update(b * bsize - self.n)

def download_url(url, output_path):
    with DownloadProgressBar(unit='B', unit_scale=True,
                             miniters=1, desc=url.split('/')[-1]) as t:
        urllib.request.urlretrieve(url, filename=output_path, reporthook=t.update_to)

def direct_file_download_open_neuro(file_list, file_types, dataset_id, dataset_version, save_dirs):
    # https://openneuro.org/crn/datasets/ds004226/snapshots/1.0.0/files/sub-001:sub-001_scans.tsv
    for i, n in enumerate(file_list):
        subject = n.split('_')[0]
        download_link = 'https://openneuro.org/crn/datasets/{}/snapshots/{}/files/{}:{}:{}'.format(dataset_id, dataset_version, subject, file_types[i],n)
        print('Attempting download from ', download_link)
        download_url(download_link, op.join(save_dirs[i], n))
        print('Ok')
        
def get_json_from_file(fname):
    f = open(fname)
    data = json.load(f)
    f.close()
    return data
````

## Skull-Stripping

````python
def get_skull_stripped_anatomical(bids_root, preproc_root, subject_id, robust=False):
    """
    Function to perform skull-stripping (removing the skull around the brain).
    This is a simple wrapper around the brain extraction tool (BET) in FSL's suite
    It assumes data to be in the BIDS format (which we will cover in the following weeks).
    The method also saves the brain mask which was used to extract the brain.

    The brain extraction is conducted only on the T1w of the participant.

    Parameters
    ----------
    bids_root: string
        The root of the BIDS directory
    preproc_root: string
        The root of the preprocessed data, where the result of the brain extraction will be saved.
    subject_id: string
        Subject ID, the subject on which brain extraction should be conducted.
    robust: bool
        Whether to conduct robust center estimation with BET or not. Default is False.
    """
    # We perform here skull stripping (you'll learn more about it next week!).
    # For now all you need to do is that we remove the bones and flesh from the MRI to get the brain!
    subject = 'sub-{}'.format(subject_id)
    anatomical_path = op.join(bids_root, subject, 'anat', 'sub-{}_T1w.nii.gz'.format(subject_id))
    betted_brain_path = op.join(preproc_root, subject, 'anat', 'sub-{}_T1w'.format(subject_id))
    os.system('bet {} {} -m {}'.format(anatomical_path, betted_brain_path, '-R' if robust else ''))
    print("Done with BET.")

resulting_mask_path = op.join(preproc_root, 'sub-101410', 'anat', 'sub-101410_T1w_mask')
get_skull_stripped_anatomical(bids_root, preproc_root, "101410", robust = True)  # To improve the fit
````

ATTENTION: LE FICHIER SETTLED EST DANS LE PREPROC ROOT

Before merging both runs together, we want to rescale them. The reason is that different runs can have different overall signal intensities or variances, because of: scanner gain or baseline drift, physiological noise changes or small differences in acquisition or head position.

````python
import os
import numpy as np
import nibabel as nib
from nilearn.image import concat_imgs, new_img_like
import matplotlib.pyplot as plt

# Raw fMRI runs
runs_info = [
    ('sub-101410_task-motorLR', 'sub-101410_task-motorLR_bold_settled.nii.gz'),
    ('sub-101410_task-motorRL', 'sub-101410_task-motorRL_bold_settled.nii.gz')
]

##### Pourquoi mettre un run name?

# Output folder for preprocessed/scaled runs
output_dir = os.path.join(preproc_root, 'sub-101410', 'func')
os.makedirs(output_dir, exist_ok=True)

concatenated = [] #merged file

# Process each run independently
for run_name, run_file in runs_info:
    run_path = os.path.join(preproc_root, 'sub-101410', 'func', run_file)  
    img = nib.load(run_path)
    data = img.get_fdata(dtype=np.float32)  # Use float32 to save memory
    
    # Simple brain mask
    brain_mask = data > 0    
    global_var = np.var(data[brain_mask], dtype=np.float64)
    
    # Rescale to unit variance
    scaled_data = data / np.sqrt(global_var)
    del data #free memory
    
    # Merge scaled run
    new_img = new_img_like(img, scaled_data)
    concatenated.append(new_img)# merging both runs
    del scaled_data #free memory


concat_img = concat_imgs(concatenated)
concat_img.to_filename(os.path.join(output_dir, 'sub-101410_task-motor_concat.nii.gz'))

````

````python
import os
import numpy as np
import nibabel as nib
from nilearn.image import concat_imgs, new_img_like
import matplotlib.pyplot as plt

# Raw fMRI runs
runs_info = [
    ('sub-101410_task-motorLR', 'sub-101410_task-motorLR_bold_settled.nii.gz'),
    ('sub-101410_task-motorRL', 'sub-101410_task-motorRL_bold_settled.nii.gz')
]

##### Pourquoi mettre un run name?

# Output folder for preprocessed/scaled runs
output_dir = os.path.join(preproc_root, 'sub-101410', 'func')
os.makedirs(output_dir, exist_ok=True)

concatenated = [] #merged file

# Process each run independently
# Simple mean matching (no z-score)
ref_mean = None
for run_name, run_file in runs_info:
    run_path = os.path.join(preproc_root, 'sub-101410', 'func', run_file)
    img = nib.load(run_path)
    data = img.get_fdata(dtype=np.float32)
    
    brain_mask = data > 0
    global_mean = np.mean(data[brain_mask], dtype=np.float64)
    
    if ref_mean is None:
        ref_mean = global_mean  # first run as reference
    
    # Scale to same mean
    scaled_data = data * (ref_mean / global_mean)
    
    new_img = new_img_like(img, scaled_data)
    concatenated.append(new_img)


concat_img = concat_imgs(concatenated)
concat_img.to_filename(os.path.join(output_dir, 'sub-101410_task-motor_concat.nii.gz'))

````

#### Framewise Displacement

````python
import matplotlib.pyplot as plt
import nibabel as nib
import os.path as op

def load_mot_params_fsl_6_dof(path):
    return pd.read_csv(path, sep='  ', header=None, 
            engine='python', names=['Rotation x', 'Rotation y', 'Rotation z','Translation x', 'Translation y', 'Translation z'])

mot_params = load_mot_params_fsl_6_dof(op.join(preproc_root, 'sub-101410', 'func', 'sub-101410_task-motor_concat_moco.par'))

def compute_FD_power(mot_params):
    framewise_diff = mot_params.diff().iloc[1:]
    rot_params = framewise_diff[['Rotation x', 'Rotation y', 'Rotation z']]
    # Estimating displacement on a 50mm radius sphere
    # To know this one, we can remember the definition of the radian!
    # Indeed, let the radian be theta, the arc length be s and the radius be r.
    # Then theta = s / r
    # We want to determine here s, for a sphere of 50mm radius and knowing theta. Easy enough!
    
    # Another way to think about it is through the line integral along the circle.
    # Integrating from 0 to theta with radius 50 will give you, unsurprisingly, r0 theta.
    converted_rots = rot_params*50
    trans_params = framewise_diff[['Translation x', 'Translation y', 'Translation z']]
    fd = converted_rots.abs().sum(axis=1) + trans_params.abs().sum(axis=1)
    return fd

fd = compute_FD_power(mot_params).to_numpy()

#threshold = np.quantile(fd,0.75) + 1.5*(np.quantile(fd,0.75) - np.quantile(fd,0.25))
threshold = 0.38

#%matplotlib inline
plt.plot(list(range(1, fd.size+1)), fd)
plt.xlabel('Volume')
plt.ylabel('FD displacement (mm)')
plt.hlines(threshold, 0, 600,colors='black', linestyles='dashed', label='FD threshold')
plt.title('Framewise displacement (ref volume = 150)')
plt.legend()

plt.savefig('FD ref 150', bbox_inches='tight')

plt.show()
````

By observing the framewise displacement plot, we can clearly see the big jump from run LR to run RL. In order to later remove the volumes with high FD, we applied a threshold: 

````python
print(f"Indices of frames above threshold: {(np.where(fd > threshold)[0] + 1).tolist()}")
````

# 3. GLM Design Matrix

Because we removed the first two volumes of both runs and merged them, we need to do the same with the events files:

````python
import pandas as pd
import numpy as np
import os.path as op

# Load event files
LRevents = pd.read_csv(op.join(bids_root, 'sub-101410', 'func', 'sub-101410_task-motorLR_events.tsv'), sep=",")
RLevents = pd.read_csv(op.join(bids_root, 'sub-101410', 'func', 'sub-101410_task-motorRL_events.tsv'), sep=",")

TR = 0.72
n_removed = 2
time_removed = TR * n_removed  # 1.44 s

def trim_first_volumes(events, time_removed):
    """Shorten the first event and shift everything so that onset >= 0."""
    events = events.copy()
    # 1️⃣ Shorten the first event by the removed time
    events.loc[0, 'duration'] = events.loc[0, 'duration'] - time_removed
    
    # 2️⃣ Shift *only* later events earlier by time_removed
    events.loc[1:, 'onset'] = events.loc[1:, 'onset'] - time_removed
    
    # 3️⃣ Keep the first onset exactly at 0
    events.loc[0, 'onset'] = 0.0
    return events

LRevents = trim_first_volumes(LRevents, time_removed)
RLevents = trim_first_volumes(RLevents, time_removed)

# 4️⃣ Concatenate with onset offset for the second run
offset_time = (LRevents['onset'] + LRevents['duration']).max()
RLevents['onset'] = RLevents['onset'] + offset_time
events_concat = pd.concat([LRevents, RLevents], ignore_index=True)
````

````python
import os
import numpy as np
import nibabel as nib
from tqdm import trange
from sklearn.decomposition import PCA, FastICA
from scipy.stats import zscore
from nilearn.input_data import NiftiMasker
from nilearn import datasets
from nilearn.plotting import plot_stat_map
import matplotlib.pyplot as plt
import os.path as op

# --- Paths and parameters ---
data_path = op.join(preproc_root, 'sub-101410','func')
data_folder = op.join(data_path, 'other')

# List .nii files
nii_files = sorted([f for f in os.listdir(data_folder)
                    if f.endswith('.nii') or f.endswith('.nii.gz')])
print("Found files:", nii_files)

# Load example image
img_ex = nib.load(op.join(data_folder, nii_files[0]))
affine = img_ex.affine
x, y, z, t = img_ex.shape
print(f"Image shape: {x,y,z,t}")

N_subj = len(nii_files)
vols_per_file = [nib.load(op.join(data_folder, f)).shape[-1] for f in nii_files]
N_tot = sum(vols_per_file)
print(f"Total timepoints across all runs: {N_tot}")

# ----------------------------------------------------------
# 🧠 Create a brain mask (instead of using all voxels)
# ----------------------------------------------------------
# You can use nilearn’s built-in EPI mask strategy
masker = NiftiMasker(mask_strategy='epi', standardize=True, dtype='float32')
masker.fit(op.join(data_folder, nii_files[0]))  # use first file to define mask

n_voxels_masked = int(masker.mask_img_.get_fdata().sum())
print(f"Brain mask contains {n_voxels_masked} voxels.")

# ----------------------------------------------------------
# Build X using the masker (memory-safe)
# ----------------------------------------------------------
X = np.zeros((N_tot, n_voxels_masked), dtype=np.float32)
print(f"Allocating X with shape: {X.shape}")

start_idx = 0
for file, n_vol in zip(nii_files, vols_per_file):
    print(f"Loading and masking {file} ({n_vol} volumes)...")
    run_data = masker.transform(op.join(data_folder, file))  # shape (n_vol, n_voxels_masked)
    X[start_idx:start_idx + n_vol, :] = run_data
    start_idx += n_vol

print("All runs loaded into X.")
print("Final matrix shape:", X.shape)

# ----------------------------------------------------------
# Z-score across voxels (time-normalization per voxel)
# ----------------------------------------------------------
# Safer and faster:
# X = zscore(X, axis=0)
# X = np.nan_to_num(X, nan=0.0)
# print("Z-scored and cleaned data matrix.")

# ----------------------------------------------------------
# PCA to estimate explained variance
# ----------------------------------------------------------
print("Running PCA to estimate explained variance...")
max_components = min(50, X.shape[0])  # safer cap
pca = PCA(n_components=max_components, svd_solver='randomized', whiten=True)
pca.fit(X)

explained_var = pca.explained_variance_ratio_
cum_var = np.cumsum(explained_var)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(range(1, len(explained_var)+1), explained_var, 'o-')
plt.title('Scree plot (individual variance)')
plt.xlabel('Component number')
plt.ylabel('Variance ratio')

plt.subplot(1, 2, 2)
plt.plot(range(1, len(cum_var)+1), cum_var, 'o-')
plt.axhline(0.95, color='gray', linestyle=':', label='95% variance')
plt.title('Cumulative explained variance')
plt.xlabel('Number of components')
plt.ylabel('Cumulative variance')
plt.legend()
plt.tight_layout()
plt.show()

opt_n = np.argmax(cum_var >= 0.5) + 1
print(f"≈ {opt_n} components explain 50% of the variance.")

# ----------------------------------------------------------
# ICA decomposition
# ----------------------------------------------------------
nb_components = [20]  # change as needed
S_dict = {}

for n in nb_components:
    print(f"Running ICA with {n} components...")
    n2 = min(2 * n, X.shape[1])  # PCA before ICA
    H = PCA(n_components=n2, svd_solver='randomized', whiten=False).fit_transform(X)

    ica = FastICA(n_components=n, random_state=1, max_iter=300, tol=0.001)
    S_ = ica.fit_transform(H)  # time courses
    W_ = ica.components_       # spatial maps (in PCA space)

    S_dict[n] = S_
    print(f"ICA done: sources shape = {S_.shape}, weights shape = {W_.shape}")

print("✅ ICA finished successfully.")
````

````python
from nilearn.plotting import plot_stat_map

# --- Reconstruct ICA components into 3D brain maps ---
print("Reconstructing ICA components into voxel space...")

# Each ICA run can have a different number of components, so we’ll iterate
for n, S_ in S_dict.items():
    print(f"Reconstructing {n} components...")

    # Retrieve PCA and ICA matrices used in this iteration
    n2 = min(2 * n, X.shape[1])
    pca = PCA(n_components=n2, svd_solver='randomized', whiten=True)
    H = pca.fit_transform(X)
    ica = FastICA(n_components=n, random_state=1, max_iter=300, tol=0.001)
    ica.fit(H)

    # Spatial maps: (n_voxels_masked × n_components)
    mixing = ica.mixing_
    spatial_maps_masked = pca.components_.T @ mixing

    # Convert back into 3D brain space using the masker
    component_imgs = masker.inverse_transform(spatial_maps_masked.T)

    # Save to disk
    out_path = op.join(data_folder, f'ICA_components_{n}_comp.nii.gz')
    component_imgs.to_filename(out_path)
    print(f"Saved: {out_path}")

    # --- Plot each component ---
    print("Plotting components...")
    for i in range(n):
        title = f"ICA Component {i+1} (of {n})"
        display = plot_stat_map(
            masker.inverse_transform(spatial_maps_masked.T[i]),
            display_mode='ortho',
            threshold=0.002,
            title=title,
            colorbar=True,
            black_bg=True
        )
````

````python
import os
import os.path as op
import numpy as np
import nibabel as nib
from tqdm import trange
from sklearn.decomposition import IncrementalPCA, FastICA
from sklearn.utils.extmath import safe_sparse_dot
from nilearn.image import new_img_like
from nilearn.plotting import plot_stat_map

# --------- user params ----------
path_func_img = op.join(preproc_root, 'sub-101410', 'func', 'sub-101410_task-motor_concat_moco_smoothed-6mm.nii.gz')
memmap_dir = '/tmp'              # directory for temporary memmap file (change if needed)
memmap_name = op.join(memmap_dir, 'fmri_masked_memmap.dat')
n_components_ica = 20
n_pca = 2 * n_components_ica  # = 60
chunk_time = 100  # must be >= n_pca              # chunk size (timepoints per read); tune for speed/memory
dtype = np.float32
# --------------------------------

print("Opening image via nibabel (no get_fdata())")
img = nib.load(path_func_img)
affine = img.affine
shape = img.shape  # (x,y,z,t)
if len(shape) != 4:
    raise RuntimeError("Expect a 4D image")
x, y, z, n_times = shape
print("Image shape:", shape)

# -----------------------
# Step 1: compute a brain mask by streaming mean image (low-memory)
# -----------------------
print("Computing mean image in chunks to derive a brain mask...")
dataobj = img.dataobj  # array proxy, does not load whole array

# Compute mean across time without loading all volumes at once
mean_vol = np.zeros((x, y, z), dtype=np.float64)
count = 0
for t0 in range(0, n_times, chunk_time):
    t1 = min(n_times, t0 + chunk_time)
    # stacking a small chunk in memory only
    chunk = np.asarray(dataobj[..., t0:t1], dtype=np.float64)
    mean_vol += chunk.sum(axis=3)
    count += (t1 - t0)

mean_vol = mean_vol / float(count)

# simple mask: threshold a small fraction of max; adjust threshold if needed
thr = mean_vol.mean() * 0.3
mask_bool = mean_vol > thr
n_brain_voxels = mask_bool.sum()
print(f"Brain voxels (threshold {thr:.3g}):", n_brain_voxels)
if n_brain_voxels == 0:
    raise RuntimeError("Mask has zero voxels: adjust thresholding step")

# Save mask as nibabel image if wanted
mask_img = new_img_like(img, mask_bool.astype(np.uint8), affine).get_fdata()  # small

# -----------------------
# Step 2: create memmap (time x n_brain_voxels) and fill it by streaming time chunks
# -----------------------
print("Creating memmap and filling it by streaming volumes...")
# remove existing memmap to avoid shape mismatch
if os.path.exists(memmap_name):
    os.remove(memmap_name)

X_mem = np.memmap(memmap_name, dtype=dtype, mode='w+', shape=(n_times, n_brain_voxels))

# helper: indices of brain voxels (flattened)
brain_idx = np.where(mask_bool.flatten())[0]

# fill memmap chunk-by-chunk to avoid large RAM usage
for t0 in trange(0, n_times, chunk_time, desc="Filling memmap"):
    t1 = min(n_times, t0 + chunk_time)
    # read chunk: shape (x,y,z,chunklen) - small if chunklen small
    chunk = np.asarray(dataobj[..., t0:t1], dtype=dtype)  # chunk loaded into memory
    # flatten space -> (voxels, chunklen)
    chunk_flat = chunk.reshape(-1, chunk.shape[3])       # (x*y*z, chunklen)
    # select brain voxels and transpose -> (chunklen, n_brain_voxels)
    block = chunk_flat[brain_idx, :].T
    X_mem[t0:t1, :] = block

# Force to disk (flush)
X_mem.flush()
print("Memmap created: ", X_mem.shape, memmap_name)

# -----------------------
# Step 3: standardize voxel-wise in-place on memmap (zero-mean unit-var)
# -----------------------
print("Standardizing memmap in-place (voxel-wise)...")
# process voxels in chunks if number of voxels is huge; here we do by voxel-blocks
vox_chunk = 5000  # tune this to fit memory (number of voxels per block)
nvox = n_brain_voxels
for v0 in trange(0, nvox, vox_chunk, desc="Z-scoring voxels"):
    v1 = min(nvox, v0 + vox_chunk)
    block = X_mem[:, v0:v1]                  # shape (n_times, blocksize)
    # compute mean/std across time
    mean = block.mean(axis=0, keepdims=True)
    std = block.std(axis=0, keepdims=True)
    std[std == 0] = 1.0
    block[:] = (block - mean) / std
    X_mem[:, v0:v1] = block
# flush again
X_mem.flush()

# -----------------------
# Step 4: Incremental PCA on memmap in time-chunks
# -----------------------
print("Running IncrementalPCA (out-of-core)...")
ipca = IncrementalPCA(n_components=n_pca)

# IncrementalPCA expects samples x features -> (n_times, n_brain_voxels)
for t0 in trange(0, n_times, chunk_time, desc="IPCA chunks"):
    t1 = min(n_times, t0 + chunk_time)
    Xchunk = np.asarray(X_mem[t0:t1, :])   # small chunk
    ipca.partial_fit(Xchunk)

# Now transform into reduced space and store reduced matrix to a small in-RAM array
print("Transforming full memmap to PCA space (will create array of shape n_times x n_pca)...")
H_mem = np.zeros((n_times, n_pca), dtype=np.float32)
for t0 in trange(0, n_times, chunk_time, desc="IPCA transform"):
    t1 = min(n_times, t0 + chunk_time)
    Xchunk = np.asarray(X_mem[t0:t1, :])
    H_mem[t0:t1, :] = ipca.transform(Xchunk)

print("PCA reduced matrix shape:", H_mem.shape)

# -----------------------
# Step 5: ICA on small PCA matrix (can be done in memory)
# -----------------------
print("Running FastICA on small PCA output...")
ica = FastICA(n_components=n_components_ica, random_state=1, max_iter=500)
S = ica.fit_transform(H_mem)   # (n_times, n_components)
A = ica.mixing_                # (n_pca, n_components)
print("ICA done. S shape:", S.shape, "A shape:", A.shape)

# -----------------------
# Step 6: back-project to voxel space to get spatial maps
# spatial_maps = (n_brain_voxels, n_components) = PCA_components.T (voxels x n_pca) dot A (n_pca x n_components)
# PCA components matrix from IncrementalPCA: ipca.components_ has shape (n_pca, n_brain_voxels) ??? careful
# Note: sklearn IncrementalPCA stores components_ as (n_components, n_features) with features = n_brain_voxels
# We need PCA components as (n_brain_voxels, n_pca) to multiply with A
# -----------------------
print("Reconstructing spatial maps...")
# ipca.components_.shape = (n_pca, n_brain_voxels)
pca_components = ipca.components_           # (n_pca, n_brain_voxels)
spatial_maps = (pca_components.T).dot(A)    # (n_brain_voxels, n_components)
print("spatial_maps shape (voxels x components):", spatial_maps.shape)

# Create 3D NIfTI images for components using mask and inverse transform
# We'll reconstruct full-volume components (zeros outside mask)
print("Saving components as NIfTI images...")
````

````python
# Number of components for ICA (You can experiment with different values, e.g., [4, 10, 20, etc.])
from sklearn.decomposition import PCA
from sklearn.decomposition import FastICA
import numpy as np

nb_components = [4]  # other values possible but first 4 chosen because they usually capture the most variance or meaningful patterns.

S_dict = {}  # Store the sources for different numbers of components

# Check if there are NaN or Inf values before applying PCA or ICA
def check_and_clean_data(data):
    """Check and clean NaN or Inf values in the dataset."""
    if np.any(np.isnan(data)) or np.any(np.isinf(data)):
        print("Warning: NaN or Inf values detected in the data. Cleaning...")
        # Replace NaNs and Infs with zeros
        data = np.nan_to_num(data, nan=0.0, posinf=0.0, neginf=0.0)
    return data

# Clean the data before applying PCA or ICA
X = check_and_clean_data(X)

# Now proceed with PCA and ICA
for n in nb_components:
    print(f"Processing {n} components")
    
    # Step 1: Apply PCA (preprocessing for ICA)
    pca = PCA(n_components=n)  # Use fewer components for PCA (just n components)
    try:
        X_pca = pca.fit_transform(X)  # Reduce dimensionality to the first n principal components
        print("PCA transformation successful. Shape of PCA result:", X_pca.shape)
        
        # Debugging: Print the variance explained by each component
        print(f"Explained variance ratio by PCA components: {pca.explained_variance_ratio_}")
        
    except ValueError as e:
        print(f"Error during PCA: {e}")
        continue  # Skip to next component if there's a problem with PCA

    # Explicitly check for NaN or Inf after PCA
    X_pca = check_and_clean_data(X_pca)
    
    # Check if PCA has caused issues with variance (e.g., constant values)
    if np.all(np.var(X_pca, axis=0) == 0):
        print("Warning: After PCA, all components have zero variance. Skipping ICA for this configuration.")
        continue
    
    # Step 2: Run FastICA
    ica = FastICA(n_components=n, random_state=1, max_iter=200)
    try:
        S_ = ica.fit_transform(X_pca)  # S_ contains the estimated sources
        print(f"ICA completed successfully for {n} components. Shape of ICA sources:", S_.shape)
    except ValueError as e:
        print(f"Error during ICA: {e}")
        continue  # Skip to next component if there's a problem with ICA
    
    # Store the sources for later use
    S_dict[n] = S_
    
    print(f"Finished processing {n} components.")
````
